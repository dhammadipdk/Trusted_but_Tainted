## Repository Download

The following repositories are required:

- AdaFace
- MagFace
- ElasticFace
- EdgeFace

````python
# Models have been taken from here

REPOS_DIR = PROJECT_ROOT / "repos"

REPOSITORIES = {

    "AdaFace":
    "https://github.com/mk-minchul/AdaFace.git",

    "MagFace":
    "https://github.com/IrvingMeng/MagFace.git",

    "ElasticFace":
    "https://github.com/fdbtrs/ElasticFace.git",

    "edgeface":
    "https://github.com/otroshi/edgeface.git"
}

REPOS_DIR.mkdir(parents=True, exist_ok=True)
````

````python
import subprocess

for repo_name, repo_url in REPOSITORIES.items():
    repo_path = REPOS_DIR / repo_name
    if repo_path.exists():
        print(f"[SKIP] {repo_name}")
        continue

    print(f"[CLONING] {repo_name}")
    subprocess.run(["git", "clone", repo_url], cwd=str(REPOS_DIR), check=True)

print("\nAll repositories ready.")
````
